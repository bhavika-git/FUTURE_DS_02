```python
# Import Pandas so we can work with the customer churn dataset
import pandas as pd

# Load the Telco Customer Churn CSV file
# The notebook is inside the "notebooks" folder,
# so "../data/" moves one level up and then into the data folder
df = pd.read_csv("../data/Telco-Customer-Churn.csv")

# Display the first five rows to confirm that the dataset loaded correctly
df.head()
```

```python
# Check the number of rows and columns
print("Dataset shape:", df.shape)

# Display column names and data types
df.info()
```

```python
# Check missing values in each column
print("Missing values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Convert TotalCharges to numeric values
# Invalid/blank values are converted to NaN
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Check how many values became missing
print("Missing TotalCharges after conversion:", df["TotalCharges"].isnull().sum())

# Remove rows with missing TotalCharges
df = df.dropna(subset=["TotalCharges"])

print("Shape after cleaning:", df.shape)
```

```python
# Count customers in each churn category
churn_counts = df["Churn"].value_counts()

# Calculate overall churn rate
churn_rate = (df["Churn"].eq("Yes").mean()) * 100

print("Churn counts:")
print(churn_counts)

print(f"\nOverall churn rate: {churn_rate:.2f}%")
```

```python
# Calculate churn rate for each contract type
contract_churn = (
    df.groupby("Contract")["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

print("Churn rate by contract type:")
print(contract_churn.round(2))
```

```python
# Create customer tenure groups in months
bins = [-1, 12, 24, 36, 48, 60, float("inf")]
labels = ["0–12 months", "13–24 months", "25–36 months",
          "37–48 months", "49–60 months", "60+ months"]

df["TenureGroup"] = pd.cut(df["tenure"], bins=bins, labels=labels)

# Calculate churn rate for each tenure group
tenure_churn = (
    df.groupby("TenureGroup", observed=True)["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
)

print("Churn rate by tenure:")
print(tenure_churn.round(2))
```

```python
# Calculate churn rate for each internet service type
internet_churn = (
    df.groupby("InternetService")["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

print("Churn rate by Internet Service:")
print(internet_churn.round(2))
```

```python
# Calculate churn rate for each payment method
payment_churn = (
    df.groupby("PaymentMethod")["Churn"]
      .apply(lambda x: (x == "Yes").mean() * 100)
      .sort_values(ascending=False)
)

print("Churn rate by Payment Method:")
print(payment_churn.round(2))
```

```python
# Compare average monthly charges for customers who stayed vs churned
monthly_charges = df.groupby("Churn")["MonthlyCharges"].mean()

print("Average Monthly Charges by Churn Status:")
print(monthly_charges.round(2))
```

```python
# Compare average tenure for customers who stayed vs churned
average_tenure = df.groupby("Churn")["tenure"].mean()

print("Average Tenure by Churn Status:")
print(average_tenure.round(2))
```

```python
# Calculate churn rates for selected customer characteristics
segment_columns = ["SeniorCitizen", "Partner", "Dependents"]

for column in segment_columns:
    print(f"\nChurn rate by {column}:")
    
    result = (
        df.groupby(column)["Churn"]
          .apply(lambda x: (x == "Yes").mean() * 100)
    )
    
    print(result.round(2))
```

```python
# Save the cleaned dataset for Power BI and documentation
df.to_csv("../data/Telco-Customer-Churn-Cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```
